## Что такое nn.Module

В PyTorch нейросеть обычно описывается как класс, который наследуется от:

```python
torch.nn.Module
```

Внутри класса обычно есть два важных блока:

```python
__init__()
forward()
```

### `__init__`

Здесь мы создаём слои нейросети.

Например:

```python
self.layer1 = nn.Linear(4, 16)
self.layer2 = nn.Linear(16, 3)
```

### `forward`

Здесь мы описываем, как данные проходят через сеть.

Например:

```python
x = self.layer1(x)
x = torch.relu(x)
x = self.layer2(x)
return x
```

---

## Что такое Linear

`nn.Linear` — это полносвязный слой.

Он реализует формулу:

```text
Z = X @ W + b
```

То есть то, что мы уже вручную делали в NumPy.

---

## Что такое Optimizer

Optimizer обновляет параметры модели.

Вручную мы писали:

```python
w = w - lr * gradient
```

В PyTorch обычно пишут:

```python
optimizer.step()
```

Популярные оптимизаторы:

- SGD;
- Adam;
- RMSprop.

Сегодня используем:

```python
torch.optim.Adam
```
