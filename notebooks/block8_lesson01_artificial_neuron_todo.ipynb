## Цель занятия

На этом занятии мы начинаем блок глубокого обучения.

Сегодня студент должен понять главную идею:

```text
нейронная сеть состоит из простых вычислительных блоков — искусственных нейронов
```

Искусственный нейрон получает признаки, умножает их на веса, добавляет смещение и применяет функцию активации.

---

## Простая формула нейрона

```text
z = x1*w1 + x2*w2 + ... + xn*wn + b
```

Где:

| Обозначение | Смысл |
|---|---|
| x | входной признак |
| w | вес признака |
| b | bias, смещение |
| z | взвешенная сумма |
| activation(z) | функция активации |
| y_pred | выход нейрона |

---

## Пример из жизни

Допустим, нейрон должен оценить вероятность того, что студент успешно сдаст работу.

Признаки:

- сколько часов студент занимался;
- насколько хорошо выполнены задания;
- насколько сложная тема.

Каждый признак имеет свой вес.

Если вес положительный — признак помогает результату.  
Если вес отрицательный — признак мешает результату.

---

## Что такое функция активации

Функция активации превращает обычное число в полезный выход модели.

Примеры:

- Step Function — да/нет;
- Sigmoid — вероятность от 0 до 1;
- ReLU — пропускает только положительные значения.

## Ячейка 6. TODO: Sigmoid

### Алгоритм

1. Создайте функцию `sigmoid(z)`.
2. Посчитайте probability.
3. Проверьте, что вероятность от 0 до 1.

````python
# TODO: создайте функцию sigmoid(z)
def sigmoid(z):
    return 1 / (1 + np.exp(-z))


# TODO: посчитайте probability
probability = sigmoid(z_function)

# TODO: выведите вероятность
print("z:", z_function)
print("Вероятность успеха:", round(float(probability), 4))

# TODO: assert 0 <= probability <= 1
assert 0 <= probability <= 1

# TODO: assert sigmoid(0) == 0.5 или почти 0.5
assert round(sigmoid(0), 4) == 0.5
````

## Ячейка 7. TODO: ReLU и графики функций активации

### Алгоритм

1. Создайте функцию `relu(z)`.
2. Создайте диапазон `z_values`.
3. Посчитайте значения Step, Sigmoid, ReLU.
4. Постройте графики.
5. Проверьте ReLU через assert.

````python
# TODO: создайте функцию relu(z)
def relu(z):
    return np.maximum(0, z)


# TODO: создайте z_values через np.linspace
z_values = np.linspace(-5, 5, 200)

# TODO: создайте step_values
step_values = np.array([step(value) for value in z_values])
sigmoid_values = sigmoid(z_values)
relu_values = relu(z_values)
# TODO: создайте sigmoid_values
plt.figure(figsize=(8, 5))
plt.plot(z_values, step_values, label="Step")
# TODO: создайте relu_values
# TODO: постройте графики Step, Sigmoid и ReLU
plt.plot(z_values, sigmoid_values, label="Sigmoid")
plt.plot(z_values, relu_values, label="ReLu")
plt.title("Функции активации")
plt.xlabel("z")
plt.ylabel("activation(z)")
plt.grid(True)
plt.legend()
plt.show()
# TODO: assert relu(-10) == 0
assert relu(-10) == 0
# TODO: assert relu(5) == 5
assert relu(5) == 5

assert len(z_values) == len(sigmoid_values)
````

## Ячейка 8. TODO: полная функция искусственного нейрона

### Алгоритм

1. Создайте функцию `neuron_predict`.
2. Внутри используйте `weighted_sum`.
3. Примените `sigmoid`.
4. Проверьте сильного и слабого студента.

````python
# TODO: создайте функцию neuron_predict(x, w, b)
def neuron_predict(x, w, b):
    z = weighted_sum(x, w, b)
    return sigmoid(z)


# TODO: посчитайте student_probability
student_probability = neuron_predict(x, w, b)
print("Вероятность успеха студента:", round(student_probability, 4))

# TODO: создайте strong_student
strong_student = np.array([1.0, 1.0, 0.1])
# TODO: создайте weak_student
weak_student = np.array([0.1, 0.2, 0.9])

# TODO: посчитайте strong_probability и weak_probability
string_probability = neuron_predict(strong_student, w, b)
weak_probability = neuron_predict(weak_student, w, b)
# TODO: assert strong_probability > weak_probability

assert string_probability > weak_probability
assert 0 <= student_probability <= 1
````
